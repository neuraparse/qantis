# Backend Quickstart Guide

This notebook demonstrates how to use the quantum backend abstraction layer.
Supports IBM Quantum (Qiskit), D-Wave, Azure Quantum, and PennyLane.

```python
from quantum_common.types import BackendType, QuantumParadigm
from quantum_common.backends.factory import create_backend, list_available_backends

# List all available backends
available = list_available_backends()
print(f"Available backends: {[b.value for b in available]}")
```

```python
# Create a local Aer simulator
backend = create_backend(BackendType.LOCAL_AER)
print(f"Backend: {backend.name}")
print(f"Paradigm: {backend.paradigm}")
print(f"Max qubits: {backend.max_qubits}")
print(f"Available: {backend.is_available()}")
```

```python
# Execute a simple Bell state circuit
from qiskit import QuantumCircuit
from quantum_common.backends.base import ExecutionRequest

qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])
qc.draw('mpl')
```

```python
# Execute on simulator
request = ExecutionRequest(circuits=[qc], shots=1024)
result = backend.execute(request)

print(f"Counts: {result.counts[0]}")
print(f"Execution time: {result.execution_time_s:.4f}s")
print(f"Quasi-distributions: {result.quasi_dists[0]}")
```

```python
# D-Wave simulator (annealing)
try:
    dwave_backend = create_backend(BackendType.LOCAL_DWAVE_SIM)
    print(f"D-Wave sim: {dwave_backend.name}")
    print(f"Paradigm: {dwave_backend.paradigm}")
    
    import dimod
    bqm = dimod.BinaryQuadraticModel({'x0': -1, 'x1': -1}, {('x0', 'x1'): 2}, 0, dimod.BINARY)
    result = dwave_backend.execute(ExecutionRequest(circuits=[bqm], options={'num_reads': 100}))
    print(f"D-Wave result: {result.counts[0]}")
except Exception as e:
    print(f"D-Wave not available: {e}")
```
