# Error Mitigation Demo

Demonstrates the composable error mitigation pipeline:
- Zero-Noise Extrapolation (ZNE)
- Readout Error Mitigation
- Probabilistic Error Cancellation (PEC)

```python
from quantum_common.mitigation.pipeline import MitigationPipeline
from quantum_common.mitigation.zne import ZNEStrategy
from quantum_common.mitigation.readout import ReadoutMitigationStrategy
from quantum_common.mitigation.pec import PECStrategy

# Build composable pipeline
pipeline = MitigationPipeline()
pipeline.add(ReadoutMitigationStrategy(calibration_shots=4096))
pipeline.add(ZNEStrategy(scale_factors=[1.0, 2.0, 3.0], factory_type='Richardson'))

print(f"Pipeline has {len(pipeline)} strategies")
```

```python
# Simulate noisy counts
raw_counts = {'00': 450, '01': 30, '10': 25, '11': 495}
print(f"Raw counts: {raw_counts}")

# Apply mitigation
mitigated = pipeline.apply(None, None, raw_counts, shots=1000)
print(f"Mitigated counts: {mitigated}")
```

```python
# Noise profile analysis
from quantum_common.mitigation.noise import NoiseProfile

profile = NoiseProfile(
    backend_name='simulated',
    single_qubit_error=0.001,
    two_qubit_error=0.01,
    readout_error=0.02,
    t1_us=300,
    t2_us=200,
)
print(f"Estimated fidelity: {profile.estimated_circuit_fidelity:.4f}")
print(f"Suggested mitigation: {profile.suggest_mitigation()}")
```
